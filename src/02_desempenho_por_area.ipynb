## Calcular e reconciliar

DuckDB consulta uma área por vez com 256 MB/uma thread. Média, Q1, mediana e Q3 usam apenas a população elegível; quartis contínuos interpolam linearmente. O CSV mantém precisão, a tabela arredonda somente a exibição. Nenhuma divisão por dez ou filtro conjunto entre áreas.

```python
resultados, relatorio = executar_desempenho(raiz)
tabela = pd.DataFrame(resultados)
colunas = ["area", "elegiveis", "media", "q1", "mediana", "q3", "zeros_elegiveis"]
display(tabela[colunas].round(2))
print("Base:", relatorio["total_base"], "| Validação:", relatorio["validacao"])
print("Hash da fonte preservado:", relatorio["sha256_fonte_antes"] == relatorio["sha256_fonte_depois"])
```

## Participação e completude

Percentuais de categorias usam todos os registros como denominador. Completude dos presentes usa somente presentes como denominador; completude da base usa o total da trusted. Eliminados são uma categoria separada, e ausência não significa nota zero.

```python
display(tabela[["area", "total_base", "presentes", "ausentes", "eliminados", "presentes_sem_nota", "pct_completude_base", "pct_completude_presentes"]].round(2))
arquivos = gerar_graficos(resultados, raiz / "reports/graficos")
for arquivo in arquivos:
    display(Image(filename=str(arquivo)))
```
