```python
import duckdb
import pandas as pd
```

```python
con = duckdb.connect(database=':memory:')
```

```python
query = """
SELECT 
    'Matemática' AS area,
    850.5 AS nota_maxima,
    450.0 AS nota_minima
UNION ALL
SELECT 
    'Linguagens' AS area,
    780.2 AS nota_maxima,
    410.0 AS nota_minima;
"""
```

```python
# 3. Retorna o resultado SQL direto como um DataFrame do Pandas
df = con.execute(query).df()
```

```python
# 4. Exibe o resultado
print("✅ Conexão com DuckDB e Pandas realizada com sucesso!\n")
display(df)
```
